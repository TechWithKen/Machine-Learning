```python
match_dataframe.dropna(subset=["goal"])


matches_with_goals = match_dataframe.dropna(subset=["goal"])
matches_with_goals["goal"].sample(20)
matches_with_goals
import xml.etree.ElementTree as ET

def extract_goals(goal_xml):
    """
    Parse the <goal> XML string and return a list of goal events.
    Handles matches with no goals (goal_xml=None).
    """
    goal_events = []
    
    if not goal_xml or pd.isna(goal_xml):
        return goal_events  # empty list if no goals
    
    root = ET.fromstring(goal_xml)
    
    for value in root.findall("value"):
        scorer = value.find("player1").text if value.find("player1") is not None else None
        assister = value.find("player2").text if value.find("player2") is not None else None
        minute = value.find("elapsed").text if value.find("elapsed") is not None else None
        goal_type = value.find("goal_type").text if value.find("goal_type") is not None else None
        subtype = value.find("subtype").text if value.find("subtype") is not None else None
        team_id = value.find("team").text if value.find("team") is not None else None
        
        goal_events.append({
            "scorer": scorer,
            "team_api_id": team_id,  
            "assister": assister,
            # "minute": minute,
            # "goal_type": goal_type,
            # "subtype": subtype,
        })
            
        
    return goal_events


matches_with_goals["Goal Information"] = match_dataframe["goal"].apply(extract_goals)
matches_with_goals = matches_with_goals.explode("Goal Information")
matches_with_goals
goals_flat = pd.concat(
    [
        matches_with_goals.drop(columns=["Goal Information"]),
        matches_with_goals["Goal Information"].apply(pd.Series),
    ],
    axis=1
)
goals_flat
goals_flat["team_api_id"] = pd.to_numeric(goals_flat["team_api_id"])
goals_flat = pd.merge(goals_flat, team_dataframe, on="team_api_id", how="left")

assists_dataframe = goals_flat.copy()
assists_dataframe.dropna(subset=["assister"], inplace=True)
assists_dataframe
goals_flat
# #goals_flat
# assists_dataframe
```

```python
goals_flat

total_goals = goals_flat.groupby("scorer")["team_long_name"].count().reset_index(name="goals")
team_count = goals_flat.groupby("scorer")["team_long_name"].nunique().reset_index(name="teams_played_for")

player_summary = total_goals.merge(team_count, on="scorer")
team_list = goals_flat.groupby("scorer")["team_long_name"].unique().reset_index(name="teams")

team_list = pd.merge(team_list, player_summary, on="scorer")
team_list.sort_values(by="goals", ascending=False)


player_assists = assists_dataframe.groupby("assister")["team_long_name"].count().reset_index(name="assists")
player_assists.sort_values(by="assists", ascending=False)

team_list.rename(columns={"scorer": "player_api_id"}, inplace=True)
player_assists.rename(columns={"assister": "player_api_id"}, inplace=True)

goals_assists = pd.merge(team_list, player_assists, on="player_api_id", how="outer").fillna(0)

goals_assists.drop(columns=["teams_played_for"], inplace=True)

goals_assists["Goals and Assists"] = goals_assists["goals"]  + goals_assists["assists"].astype("int64")

goals_assists.sort_values(by="Goals and Assists", ascending=False)

new_dataframe = player_dataframe[["player_api_id", "player_name"]]

new_dataframe["player_api_id"] = new_dataframe["player_api_id"].astype("int64")
goals_assists["player_api_id"] = goals_assists["player_api_id"].astype("int64")
new_dataframe = pd.merge(new_dataframe, goals_assists, on="player_api_id", how="left").fillna(0)


new_dataframe.sort_values(by="Goals and Assists", ascending=False).head(40)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve
from sklearn.linear_model import LogisticRegression



```

```python
cuisine_dataset = pd.read_csv("./Balanced_Cuisine.csv")
cuisine_label = cuisine_dataset["cuisine"]
cuisine_dataset.drop(columns=["cuisine"], inplace=True)

X,y = cuisine_dataset, cuisine_label
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=20)


```

```python
model = LogisticRegression(max_iter=1000, class_weight="balanced", C=10)

model.fit(X_train, y_train)
result = model.predict(X_test)
model.score(X_test, y_test)
confusion_matrix(y_test, result)

```
